```python
import matplotlib.pyplot as plt
```

# line Chart

```python
x=[1,2,3,4,5]
y=[200,300,400,500,600]
```

```python
plt.plot(x,y,color="blue")
plt.title("Sales report")
plt.xlabel("days")
plt.ylabel("sales")
plt.show()
```

# Bar Graph

```python
x=[1,2,3,4,5]
y=["Monday","Tuesday","Wednesday","Thursday","Friday"]
```

```python
plt.bar(x,y )
plt.title("Sales per Day")
plt.show()
```

# pie chart

```python
x=[1,2,3,4,5]
plt.pie(x, autopct="%1.1f%%")
plt.show()
```

# Histogram

```python
x=[1,2,3,4,5]
plt.hist(x, bins =4)
plt.show()
```

# Scatter Plot

```python
x=[1,2,3,4,5]
y=[100,200,300,400,500]
plt.scatter(x,y)
plt.show()
```

# Area chart

```python
x=[1,2,3,4,5]
y=[100,200,300,400,500]
plt.fill_between(x,y)
plt.show()
```

# Stacked chart

```python
x=[1,2,3,4,5]
y1=[4,3,2,3,5]
y2=[3,5,7,8,2]
plt.stackplot(x,y1,y2)
plt.show()
```
